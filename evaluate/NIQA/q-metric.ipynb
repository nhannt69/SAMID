```python
import numpy as np
from skimage.util import view_as_windows
from numpy.linalg import svd
```

```python
# compute local image gradient
# i use sobel filter

from scipy.ndimage import sobel

def compute_gradient(image):
    gx = sobel(image, axis = 0)
    gy = sobel(image, axis = 1)
    return gx, gy
```

```python
#extract local patches
def extract_patches(image, patch_size):
    """
    Extract patches from the image.
    """
    patches = view_as_windows(image, (patch_size, patch_size))
    return patches
```

```python
# transform patches to gradient matrix
def patch_gradient_matrix(gx_patch, gy_patch):
    # Flatten and stack gradients
    return np.stack([gx_patch.flatten(), gy_patch.flatten()], axis=1)
```

```python
#calcuate the largest singular value for each patch
def compute_largest_singular_value(patches):
    U, S, Vt = svd(patches, full_matrices=False)
    return S[0]  # Return the largest singular value
```

```python
def compute_q_metric(image, patch_size=5):
    gx, gy = compute_gradient(image)
    patches_x = extract_patches(gx, patch_size)
    patches_y = extract_patches(gy, patch_size)
    q_values = []
    for i in range(patches_x.shape[0]):
        for j in range(patches_x.shape[1]):
            grad_matrix = patch_gradient_matrix(patches_x[i, j], patches_y[i, j])
            q = compute_largest_singular_value(grad_matrix)
            q_values.append(q)
    return np.mean(q_values)
```

```python
from skimage import io, color, img_as_float
from glob import glob
from tqdm import tqdm

for img_path in tqdm(glob('../../data/unsupervised/*.png')):
    img = img_as_float(color.rgb2gray(io.imread(img_path)))
    q_score = compute_q_metric(img, patch_size=900)
    print(f"Q-metric score for {img_path}: {q_score}")
```
