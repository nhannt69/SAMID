```python
import imageio
import numpy as np
import os
import h5py

def convert_srgb_mat_to_png(data_folder, out_folder):
    '''
    Convert each full sRGB image in DND from .mat to .png without cropping.

    data_folder : folder where 'images_srgb' and 'info.mat' are located
    out_folder  : folder where output PNGs will be saved
    '''
    os.makedirs(out_folder, exist_ok=True)

    for i in range(50):
        mat_path = os.path.join(data_folder, 'images_srgb', f'{i+1:04d}.mat')
        with h5py.File(mat_path, 'r') as img_file:
            Inoisy = np.float32(np.array(img_file['InoisySRGB']).T)

        # Convert to uint8 image
        Inoisy = np.clip(Inoisy, 0, 1)
        img_uint8 = (Inoisy * 255).astype(np.uint8)

        # Save to PNG
        out_path = os.path.join(out_folder, f'{i+1:04d}.png')
        imageio.imwrite(out_path, img_uint8)

        print(f"[{i+1}/50] Converted and saved: {out_path}")
        

    print("✅ Done converting all sRGB .mat images to .png format.")

# Example usage:
convert_srgb_mat_to_png('data/benchmark', 'data/benchmark/srgb_full_png')
```
