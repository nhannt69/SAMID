```python
import os
from PIL import Image
import numpy as np

def extract_patches(image, patch_size=256, overlap=0):
    stride = patch_size - overlap
    patches = []

    img_h, img_w = image.shape[:2]
    for y in range(0, img_h - patch_size + 1, stride):
        for x in range(0, img_w - patch_size + 1, stride):
            patch = image[y:y + patch_size, x:x + patch_size]
            patches.append(patch)
    return patches

def process_folder(input_folder, output_folder, patch_size=512, overlap=0):
    os.makedirs(output_folder, exist_ok=True)

    image_extensions = (".png", ".jpg", ".jpeg", ".tif", ".bmp")
    image_files = [f for f in os.listdir(input_folder) if f.lower().endswith(image_extensions)]

    for img_file in image_files:
        img_path = os.path.join(input_folder, img_file)
        img = Image.open(img_path).convert("RGB")
        img_np = np.array(img)

        patches = extract_patches(img_np, patch_size, overlap)

        base_name = os.path.splitext(img_file)[0]
        for i, patch in enumerate(patches):
            patch_img = Image.fromarray(patch)
            patch_filename = f"{base_name}_patch_{i}.png"
            patch_path = os.path.join(output_folder, patch_filename)
   
            patch_img.save(patch_path)

        print(f"Processed {img_file}: {len(patches)} patches saved.")

# Example usage
input_folder = "SAMCycle/Test"
output_folder = "SAMCycle/testA"
process_folder(input_folder, output_folder, patch_size=256, overlap=0)
```
