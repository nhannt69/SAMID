```python
import os
import cv2

# Folder with noisy grayscale images
input_folder = r'D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\experimental\degradation\\test'

# Folder to save filtered (denoised) images
output_folder = os.path.join(input_folder, 'filtered')
os.makedirs(output_folder, exist_ok=True)

# Gaussian filter parameters
kernel_size = (5, 5)  # You can adjust: (3,3), (7,7), etc.
sigma = 0             # 0 lets OpenCV choose based on kernel size

# Apply Gaussian filter to all images
for filename in os.listdir(input_folder):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tif', '.tiff')):
        img_path = os.path.join(input_folder, filename)
        image = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

        if image is None:
            print(f"⚠️ Skipping unreadable file: {filename}")
            continue

        # Apply Gaussian blur
        filtered = cv2.GaussianBlur(image, kernel_size, sigma)

        # Save result
        output_path = os.path.join(output_folder, filename)
        cv2.imwrite(output_path, filtered)

print("✅ Done filtering all images with Gaussian blur.")
```

```python
# Folder containing the files to rename
folder_path = r'D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\BM3D'
import os
# # Rename files by removing 'denoised_' from their names
for filename in os.listdir(folder_path):
    if filename.startswith('basic_'):
        old_path = os.path.join(folder_path, filename)
        new_filename = filename.replace('basic_', '', 1)  # Remove 'denoised_' only once
        new_path = os.path.join(folder_path, new_filename)
        os.rename(old_path, new_path)

print("✅ Renaming completed.")
```

```python
# Remove '_SwinIR' from filenames in the folder
for filename in os.listdir(folder_path):
    if '_SwinIR' in filename:
        old_path = os.path.join(folder_path, filename)
        new_filename = filename.replace('_SwinIR', '')  # Remove '_SwinIR'
        new_path = os.path.join(folder_path, new_filename)
        os.rename(old_path, new_path)

print("✅ '_SwinIR' removed from filenames.")
```

```python
#extract to png from tiff

from glob import glob
from PIL import Image
import os

for tiff_path in glob(r'D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\dataset\\private_dataset\\tiff\denoise\\*.tiff'):
    print(f"Processing TIFF file: {tiff_path}")

    # Output folder to save PNGs
    output_dir = os.path.join(tiff_path.split('.')[0], 'extracted_pngs')
    os.makedirs(output_dir, exist_ok=True)

    # Open the TIFF image
    with Image.open(f'{tiff_path}') as img:
        for i in range(img.n_frames):
            img.seek(i)  # Go to each frame
            img.save(f"{output_dir}/{os.path.basename(tiff_path).split('.')[0]}_{i+1}.png", format="PNG")

    print(f"Conversion complete! PNGs saved in '{output_dir}'")
```
