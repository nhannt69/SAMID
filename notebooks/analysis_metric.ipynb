## Mapping between iqa table and noise ratio table (each method)

```python
import pandas as pd
from glob import glob
import os

# Define the folder and output file paths
input_folder = "D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\Metric\\NoiseRatio"
output_file = "D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\Metric\\iqa_2.xlsx"

# Initialize an empty DataFrame to store combined data
combined_data = pd.DataFrame()

# Iterate through all Excel files in the folder
for file_path in glob(os.path.join(input_folder, "*.xlsx")):
    # Extract the method name from the file name
    method_name = os.path.basename(file_path).split("_")[0]
    
    # Read the current file
    df = pd.read_excel(file_path)
    
    # Add a column for the method name
    df["Method"] = method_name
    
    # Append the data to the combined DataFrame
    combined_data = pd.concat([combined_data, df], ignore_index=True)

# Save the combined data to the output file
# combined_data.to_excel(output_file, index=False)
print(combined_data)
```

```python
import pandas as pd

# # Step 1: Original long-form table (simplified to just one noise ratio = 0.050 for illustration)
# data = pd.read_excel("D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\Metric\\iqa_2.xlsx")

# # Step 2: Convert to DataFrame
# df = pd.DataFrame(data)

# Step 3: Melt the DataFrame to long format
df_melted = combined_data.melt(id_vars=["Noise Ratio", "Method"], 
                    value_vars=["PSNR", "SSIM", "LPIPS-Alex", "LPIPS-VGG", "CLIPIQA", "MUSIQ", "FID"],
                    var_name="Metric", value_name="Value")

# Step 4: Pivot the table
df_pivot = df_melted.pivot_table(index=["Noise Ratio", "Metric"], columns="Method", values="Value").reset_index()

# Optional: Rename LPIPS metrics
df_pivot["Metric"] = df_pivot["Metric"].replace({
    "LPIPS-Alex": "LPIPS (Alex)",
    "LPIPS-VGG": "LPIPS (VGG-16)"
})

# Display the result
# print(df_pivot)
desired_order = [
    "PSNR",
    "SSIM",
    "LPIPS (VGG-16)",
    "LPIPS (Alex)",
    "MUSIQ",
    "CLIPIQA",
    "FID"
]

# Apply the custom row ordering
df_ordered = df_pivot.set_index("Metric").loc[desired_order].reset_index()

# Optional: Sort by Noise Ratio first if there are multiple noise levels
df_ordered = df_ordered.sort_values(by=["Noise Ratio", "Metric"])

# Display result
print(df_ordered)
```

```python
# Analysis metric 

import pandas as pd
import numpy as np

# Load Excel file
file_path = r"D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\Metric\\iqa.xlsx"
df = pd.read_excel(file_path, sheet_name="Raw")

# Define metric rules
higher_better = {"PSNR", "SSIM", "CLIPIQA"}
lower_better = {"FID", "LPIPS (Alex)", "LPIPS (VGG-16)", "MUSIQ"}

# Style function: row-wise top 3 coloring
def highlight_top3_row(row):
    metric = row['Metric']
    if metric not in higher_better and metric not in lower_better:
        return [''] * len(row)

    values = row.iloc[2:].astype(float)
    ascending = metric in lower_better  # True for FID, LPIPS, etc.

    sorted_idx = values.sort_values(ascending=ascending).index
    styles = [''] * len(row)

    # Build style color list (first red, next two blue)
    for i, col in enumerate(row.index):
        if col in sorted_idx[:1]:
            styles[i] = 'color: red'
        elif col in sorted_idx[1:2]:
            styles[i] = 'color: blue'
    return styles

# Apply style
styled = df.style.apply(highlight_top3_row, axis=1)

# Save to Excel
output_path = r"D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\Metric\\styled_iqa_top3_per_row_.xlsx"
styled.to_excel(output_path, index=False, engine='openpyxl')

print("✅ Top-3 per-row styling complete. Saved to:", output_path)
```

## Mapping by ImageName

```python
import pandas as pd
import os

# Folder containing Excel files
folder_path = r'D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Unsupervised\\Metric\\EXCEL'

# File list from image
file_names = [
    'adnet_niqa.xlsx',
    'Blur_niqa.xlsx',
    'bm3d_niqa.xlsx',
    'CBDNET_niqa.xlsx',
    'DNCNN_niqa.xlsx',
    'FDNCNN_niqa.xlsx',
    'FFDNet_niqa.xlsx',
    'input_niqa.xlsx',
    'InvDN_niqa.xlsx',
    'IRCNet_niqa.xlsx',
    'output_niqa.xlsx',
    'Restormer_niqa.xlsx',
    'SwinIR_niqa.xlsx'
]

# Initialize an empty list to hold individual model metric DataFrames
merged_by_model = []

for file in file_names:
    print(file)
    model_name = file.replace('_niqa.xlsx', '').strip()  # Remove suffix and whitespace
    file_path = os.path.join(folder_path, file)

    # Read Excel file
    df = pd.read_excel(file_path)

    # Remove whitespace from column names just in case
    df.columns = df.columns.str.strip()

    # Filter only ImageName that contains "_layer_7"
    df = df[df['ImageName'].str.contains('layer_7', na=False)]

    # Add model prefix to each metric column
    metric_cols = ['brisque', 'niqe', 'piqe', 'MetaIQA', 'RankIQA']
    df = df[['ImageName'] + metric_cols].copy()
    df = df.rename(columns={col: f"{model_name}_{col}" for col in metric_cols})

    merged_by_model.append(df)

# Merge all DataFrames on 'ImageName'
from functools import reduce
final_df = reduce(lambda left, right: pd.merge(left, right, on='ImageName', how='outer'), merged_by_model)

# Save result
output_path = os.path.join(folder_path, 'merged_layer7_metrics.xlsx')
final_df.to_excel(output_path, index=False)

print("Merged and filtered Excel saved to:", output_path)
```

```python
import pandas as pd
import os

# Load merged Excel file
input_path = r'D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Unsupervised\\Metric\\EXCEL\\merged_layer7_metrics.xlsx'
df = pd.read_excel(input_path)

# Extract method names from column headers (e.g., "adnet_brisque" → "adnet")
metric_types = ['brisque', 'niqe', 'piqe', 'MetaIQA', 'RankIQA']
methods = sorted({col.split('_')[0] for col in df.columns if '_' in col and any(m in col for m in metric_types)})

# Build long format
long_rows = []
for _, row in df.iterrows():
    image = row['ImageName']
    for method in methods:
        entry = {'ImageName': image, 'Method': method}
        for metric in metric_types:
            col_name = f"{method}_{metric}"
            if col_name in row:
                entry[metric] = row[col_name]
            else:
                entry[metric] = None  # In case missing
        long_rows.append(entry)

# Convert to DataFrame
long_df = pd.DataFrame(long_rows)

# Save to Excel
output_path = input_path.replace('.xlsx', '_long_format.xlsx')
long_df.to_excel(output_path, index=False)

print("Long format Excel saved to:", output_path)
```

```python

import pandas as pd
import numpy as np
from glob import glob
import os

# Path to your Excel files
input_folder = r"D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Unsupervised\\Metric\\EXCEL"
output_path = os.path.join(input_folder, "layer7_average_metrics.xlsx")

summary_list = []

for excel_path in glob(os.path.join(input_folder, "*.xlsx")):
    method_name = os.path.splitext(os.path.basename(excel_path))[0]

    try:
        df = pd.read_excel(excel_path, engine='openpyxl')  # Explicit engine
    except Exception as e:
        print(f"Error reading {excel_path}: {e}")
        continue

    if 'ImageName' not in df.columns:
        print(f"Skipping {method_name}: 'ImageName' column not found.")
        continue

    df_layer_7 = df[df['ImageName'].str.contains('layer_7', na=False)]

    if df_layer_7.empty:
        print(f"No 'layer_7' rows in {method_name}")
        continue

    avg_metrics = df_layer_7[['brisque', 'niqe', 'piqe', 'MetaIQA', 'RankIQA']].mean()

    summary_list.append({
        "Method": method_name,
        "brisque": avg_metrics.get('brisque', np.nan),
        "niqe": avg_metrics.get('niqe', np.nan),
        "piqe": avg_metrics.get('piqe', np.nan),
        "MetaIQA": avg_metrics.get('MetaIQA', np.nan),
        "RankIQA": avg_metrics.get('RankIQA', np.nan),
    })

summary_df = pd.DataFrame(summary_list)

# Save results
summary_df.to_excel(output_path, index=False)
print("Saved average layer_7 metrics to:", output_path)
```

```python
import pandas as pd
import re

# Read the file
with open("D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\FFDNet\\metric.log", "r") as file:
    lines = file.readlines()

# Extract data rows using regex
data = []
pattern = re.compile(
    r"(Top_.*?\.png)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)\s*\|\s*([\d.]+)"
)

for line in lines:
    match = pattern.search(line)
    if match:
        row = [match.group(i) for i in range(1, 9)]
        # Convert numeric values to float
        row[1:] = list(map(float, row[1:]))
        data.append(row)

# Define column names
columns = ["Image Name", "PSNR", "SSIM", "LPIPS-VGG", "LPIPS-Alex", "CLIPIQA", "MUSIQ", "FID"]

# Create DataFrame
df = pd.DataFrame(data, columns=columns)

# Save to Excel
df.to_excel("D:\\nbmlab\\research\\paper\\resolution\\ImageDenoising\\Content\\ResultImage\\Supervised\\ffdnet_niqa2.xlsx", index=False)
```
